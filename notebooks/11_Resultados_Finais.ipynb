```python
# ------------------------------------------------------------
# Configuração do projeto (execução local — VS Code)
# Garante que a raiz do projeto esteja no sys.path, funcionando tanto se o
# diretório de trabalho do kernel for notebooks/ quanto a raiz do projeto.
# ------------------------------------------------------------
import sys
from pathlib import Path

_raiz = Path.cwd().resolve()
for _cand in (_raiz, *_raiz.parents):
    if (_cand / 'src' / 'config.py').exists():
        _raiz = _cand
        break
else:
    raise RuntimeError(
        'Não foi possível localizar a pasta src/config.py a partir de '
        f'{Path.cwd()}. Abra a pasta TCC_TEA_TDAH/ no VS Code.'
    )
if str(_raiz) not in sys.path:
    sys.path.insert(0, str(_raiz))

import os
import random
import numpy as np

from src.config import BASE_DIR, PATHS, SEED, criar_pastas
from src.utils import exigir_arquivo

criar_pastas()

random.seed(SEED)
np.random.seed(SEED)

print('Projeto (BASE_DIR) =', BASE_DIR)
```

## Carregar resultados existentes

```python
import pandas as pd
import matplotlib.pyplot as plt

arquivo = PATHS['results'] / '09_comparacao_final_todos_modelos.csv'
exigir_arquivo(arquivo, '09_Modelo_Deep_Learning.ipynb (que consolida também os resultados do 08)')

df_final = pd.read_csv(arquivo)
df_final.round(4)
```

## Tabela final (média ± desvio padrão entre os 5 folds)

```python
tabela_final = pd.DataFrame({'Modelo': df_final['modelo']})
for nome, col in [('Accuracy', 'accuracy'), ('F1', 'f1'), ('AUC-ROC', 'auc_roc')]:
    tabela_final[nome] = (df_final[f'{col}_media'].map('{:.3f}'.format) + ' ± '
                          + df_final[f'{col}_std'].map('{:.3f}'.format))

tabela_final.to_csv(PATHS['results'] / '11_tabela_resultados_finais.csv', index=False)
print('Tabela salva em:', PATHS['results'] / '11_tabela_resultados_finais.csv')
tabela_final
```

## Comparação gráfica dos modelos (Accuracy, F1, AUC-ROC)

```python
import numpy as np

fig, ax = plt.subplots(figsize=(11, 5))
x_pos = np.arange(len(df_final))
largura = 0.25

ax.bar(x_pos - largura, df_final['accuracy_media'], largura,
       yerr=df_final['accuracy_std'], label='Accuracy', capsize=4)
ax.bar(x_pos, df_final['f1_media'], largura,
       yerr=df_final['f1_std'], label='F1', capsize=4)
ax.bar(x_pos + largura, df_final['auc_roc_media'], largura,
       yerr=df_final['auc_roc_std'], label='AUC-ROC', capsize=4)

ax.set_xticks(x_pos)
ax.set_xticklabels(df_final['modelo'], rotation=15, ha='right')
ax.set_ylabel('Score')
ax.set_title('Comparação final dos modelos (ABIDE I) — média ± desvio padrão entre folds')
ax.legend()
plt.tight_layout()
plt.savefig(PATHS['figures'] / '11_comparacao_final_modelos.png', dpi=300, bbox_inches='tight')
plt.show()
```
